```python
import matplotlib.pyplot as plt


x_labels = ['Saturday', 'Sunday', 'Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday']
y = [45, 40, 35, 40, 25, 35, 15]

fig, ax = plt.subplots(figsize=(8, 6))

ax.plot(
    range(len(x_labels)),
    y,
    marker='p',
    markersize=12,
    markerfacecolor='orange',
    linestyle='dashed',
    color='oldlace',
    linewidth=2
)

ax.set_xticks(range(len(x_labels)))
ax.set_xticklabels(x_labels, color='mediumvioletred')

ax.set_xlabel('Day', fontsize=16, color='blueviolet')
ax.set_ylabel('Time (minutes)', fontsize=16, color='blueviolet')

ax.set_title('Minutes I ran last week', fontsize=18, color='mediumvioletred')

ax.set_facecolor('blueviolet')

fig.set_facecolor('oldlace')

plt.show()
```

    برای ساخته‌شدن فایل <code>result.zip</code> سلول زیر را اجرا کنید. توجه داشته باشید که پیش از اجرای سلول زیر تغییرات اعمال شده در نت‌بوک را ذخیره کرده باشید (<code>ctrl+s</code>) تا در صورت نیاز به پشتیبانی امکان بررسی کد شما وجود داشته باشد.

```python
import zlib
import zipfile
import joblib

joblib.dump(fig,"fig")
joblib.dump(fig,"fig_t")

def compress(file_names):
    print("File Paths:")
    print(file_names)
    # Select the compression mode ZIP_DEFLATED for compression
    # or zipfile.ZIP_STORED to just store the file
    compression = zipfile.ZIP_DEFLATED
    # create the zip file first parameter path/name, second mode
    with zipfile.ZipFile("result.zip", mode="w") as zf:
        for file_name in file_names:
            # Add file to the zip file
            # first parameter file to zip, second filename in zip
            zf.write('./' + file_name, file_name, compress_type=compression)


file_names = ["fig", "what_is_done_can_be_undone.ipynb"]
compress(file_names)
```
