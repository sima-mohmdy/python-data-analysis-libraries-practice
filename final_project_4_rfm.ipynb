```python
import time
from datetime import datetime, timedelta, date, time

last_day = ((df['InvoiceDate'].max()) + timedelta(1))

last_day
```

<p dir=rtl style="direction: rtl;text-align: justify;line-height:200%;font-family:vazir;font-size:medium">
<font face="vazirmatn" size=3>
    اکنون می‌توانید به ازای هر مشتری، تاریخ آخرین خرید وی را پیدا کرده و تعداد روز گذشته از <code>last_day</code> را محاسبه کنید.
    <br>
    حاصل را در یک دیتافریم به نام <code>df_customer_segments</code> و با دو ستون <code>CustomerId</code> (شناسه‌ی مشتری) و <code>Recency</code> (مقدار ماندگاری محاسبه‌شده) ذخیره کنید. همچنین دقت کنید، دیتافریم <code>df_customer_segments</code> باید بر اساس ستون <code>CustomerId</code> (شناسه‌ی مشتری) مرتب‌سازی شده باشد.

```python
last_buying_for_each_customer = df.groupby('CustomerId')['InvoiceDate'].max()
last_buying_for_each_customer
```

```python
df_customer_segments = last_day - last_buying_for_each_customer
df_customer_segments.head()
```

```python
df_customer_segments = (last_day - last_buying_for_each_customer).dt.days
df_customer_segments.head()
```

```python
df_customer_segments = df_customer_segments.reset_index()
df_customer_segments.head()
```

```python
df_customer_segments = df_customer_segments.sort_values(by='CustomerId')
df_customer_segments.head()
```

```python
df_customer_segments = df_customer_segments.rename(columns={"InvoiceDate": "Recency"})
df_customer_segments.head()
```

     حاصل را در ستونی به نام <code>MonetaryValue</code> به دیتافریم <code>df_customer_segments</code> اضافه کنید.

<h2 dir=rtl align=right style="line-height:200%;font-family:vazir;color:#0099cc">
<font face="vazirmatn" color="#0099cc">
    گروه‌بندی هر معیار
</font>
</h2>

<p dir=rtl style="direction: rtl;text-align: justify;line-height:200%;font-family:vazir;font-size:medium">
<font face="vazirmatn" size=3>
    اکنون نیاز است هرکدام از این سه معیار را براساس چارک اول،‌ دوم و سوم به شکل زیر به ۴ گروه مختلف تقسیم کنید
    (برای یادآوری مفهوم چارک،
    <a href="https://fa.wikipedia.org/wiki/%DA%86%D8%A7%D8%B1%DA%A9_(%D8%A2%D9%85%D8%A7%D8%B1)">این مطلب</a>
    را از ویکی‌پدیای فارسی بخوانید).
</font>
</p>

<center>
<div dir=rtl style="direction: rtl;line-height:200%;font-family:vazir;font-size:medium">
<font face="vazirmatn" size=3>

| <b>شماره گروه</b> | <b>شرط</b> |
| :---: | :---: |
| 1 | <code dir=ltr>value <= Q1</code> |
| 2 | <code dir=ltr>Q1 < value <= Q2</code> |
| 3 | <code dir=ltr>Q2 < value <= Q3</code> |
| 4 | <code dir=ltr>Q3 < value</code> |

</font>
</div>
</center>

<p dir=rtl style="direction: rtl;text-align: justify;line-height:200%;font-family:vazir;font-size:medium">
<font face="vazirmatn" size=3>
    به ازای هر کدام از این سه معیار چنین کاری را انجام داده و نتایج را در سه ستون دیگر به نام‌های
    <code>F_quartile</code>، <code>R_quartile</code> و <code>M_quartile</code>
    ذخیره کنید.
    توجه کنید که مقادیر این ستون‌ها معادل شماره گروه یعنی اعداد <code>1</code> تا <code>4</code> خواهد بود.
</font>
</p>

```python
Q1_Recency = df_customer_segments['Recency'].quantile(0.25)
Q2_Recency = df_customer_segments['Recency'].quantile(0.5)
Q3_Recency = df_customer_segments['Recency'].quantile(0.75)

Q1_Frequency = df_customer_segments['Frequency'].quantile(0.25)
Q2_Frequency = df_customer_segments['Frequency'].quantile(0.5)
Q3_Frequency = df_customer_segments['Frequency'].quantile(0.75)

Q1_MonetaryValue = df_customer_segments['MonetaryValue'].quantile(0.25)
Q2_MonetaryValue = df_customer_segments['MonetaryValue'].quantile(0.5)
Q3_MonetaryValue = df_customer_segments['MonetaryValue'].quantile(0.75)
```

```python
#defining function filter
def filter_recency(value):
    if value <= Q1_Recency :
        return 1
    elif (Q1_Recency  < value <= Q2_Recency):
        return 2
    elif (Q2_Recency < value <= Q3_Recency):
        return 3
    elif (Q3_Recency < value):
      return 4
#applying the filter function to 'R_quartile' column
df_customer_segments['R_quartile'] = df_customer_segments['Recency'].apply(filter_recency)

##########################################################################################

#defining function filter
def filter_frequency(value):
    if value <= Q1_Frequency :
        return 1
    elif (Q1_Frequency  < value <= Q2_Frequency):
        return 2
    elif (Q2_Frequency < value <= Q3_Frequency):
        return 3
    elif (Q3_Frequency < value):
      return 4
#applying the filter function to 'F_quartile' column
df_customer_segments['F_quartile'] = df_customer_segments['Frequency'].apply(filter_frequency)

###########################################################################################

#defining function filter
def filter_monetary(value):
    if value <= Q1_MonetaryValue :
        return 1
    elif (Q1_MonetaryValue  < value <= Q2_MonetaryValue):
        return 2
    elif (Q2_MonetaryValue < value <= Q3_MonetaryValue):
        return 3
    elif (Q3_MonetaryValue < value):
      return 4
#applying the filter function to 'M_quartile' column
df_customer_segments['M_quartile'] = df_customer_segments['MonetaryValue'].apply(filter_monetary)
```

<font face="vazirmatn" color="#0099cc">
    معیار RFM
</font>
</h2>

<p dir=rtl style="direction: rtl;text-align: justify;line-height:200%;font-family:vazir;font-size:medium">
<font face="vazirmatn" size=3>
    اکنون می‌توانید با کنار هم قرار دادن مقادیر ستون‌های
    <code dir=ltr>[R_quartile, F_quartile, M_quartile]</code>
    مقدار نهایی <i>RFM</i> را محاسبه کرده و در ستونی به همین نام ذخیره کنید.
    به‌‌عنوان مثال اگر مقادیر این ستون‌ها برای یک مشتری برابر
    <code dir=ltr>[1, 2, 3]</code> بود،
    آن‌گاه مقدار نهایی <i>RFM</i> برابر با <code>123</code> است.
</font>
</p>

```python
df_customer_segments['RFM'] = df_customer_segments.apply(lambda row: str(row['R_quartile']) + str(row['F_quartile'])+ str(row['M_quartile']), axis=1)
df_customer_segments.head()
```

<h2 dir=rtl align=right style="line-height:200%;font-family:vazir;color:#0099cc">
<font face="vazirmatn" color="#0099cc">
    گروه‌بندی مشتریان
</font>
</h2>

<p dir=rtl style="direction: rtl;text-align: justify;line-height:200%;font-family:vazir;font-size:medium">
<font face="vazirmatn" size=3>
    حال باید براساس مقادیر ستون <code>RFM</code> و طبق جدول زیر به گروه‌بندی مشتریان بپردازید و نام گروه آن‌ها را در ستونی به نام <code>Segment</code> ذخیره کنید.
    توجه داشته باشید که در این جدول، <code>X</code> به معنی هر مقداری است. به‌عنوان مثال <code>14X</code> به معنی <code>141</code>، <code>142</code>، <code>143</code> و <code>144</code> است.
</font>
</p>

<center>
<div dir=rtl style="direction: rtl;line-height:200%;font-family:vazir;font-size:medium">
<font face="vazirmatn" size=3>

| <b>Segment</b> | <b>RFM</b> |
| :---: | :---: |
| Best | <code dir=ltr>144</code> |
| AlmostLost | <code dir=ltr>344</code> |
| LostBigSpenders | <code dir=ltr>444</code> |
| LostCheap | <code dir=ltr>441</code> |
| Loyal | <code dir=ltr>X4X</code> |
| BigSpenders | <code dir=ltr>XX4</code> |

</font>
</div>
</center>

<p dir=rtl style="direction: rtl;text-align: justify;line-height:200%;font-family:vazir;font-size:medium">
<font face="vazirmatn" size=3>
    <span style="color:purple"><b>نکته ۱:</b></span>
    در صورتی‌که یک مشتری در هیچ‌یک از گروه‌های جدول قرار نگرفت، گروه او را Normal اعلام کنید.
    <br>
    <span style="color:purple"><b>نکته ۲:</b></span>
    همچنین اگر یک RFM می‌تواند در دو گروه از جدول قرار بگیرد، اولویت با گروهی‌ست که در جدول، <b>بالاتر</b> آمده باشد.
</font>
</p>

```python
#defining function filter
def filter(value):
    if value == '144' :
        return 'Best'
    elif value == '344':
        return 'AlmostLost'
    elif value == '444':
        return 'LostBigSpenders'
    elif value == '441':
      return 'LostCheap'

    #elif value == 'X4X':
    elif value[1] == '4':
      return 'Loyal'

    #elif value == 'XX4':
    elif value[2] == '4':
      return 'BigSpenders'
    else:
      return 'Normal'

#applying the filter function to 'Segment' column
df_customer_segments['Segment'] = df_customer_segments['RFM'].apply(filter)
df_customer_segments.head()
```
